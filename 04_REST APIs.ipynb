## How a browser retrieves a web page

```text
 GET      https://   mysite.com   /path/to/page.html
 └───┘    └─────┘   └────────┘   └────────────────┘
method    protocol      domain       resource path
 verb
                    └─────────────────────────────┘
                    URI (Uniform Resource Identifier)
          └───────────────────────────────────────┘
                 URL (Uniform Resource Locator)
```

→ also supports `query arguments`

&nbsp;&nbsp;&nbsp;&nbsp;→ basically like named arguments in Python functions

```text
GET https://mysite.com/currentTemp?city=Chicago&units=metric
                                   └──────────┘ └───────────┘
                                     query          query
                                   argument       argument
```

→ web server at `mysite.com` waits to receive these requests

→ browser sends request to web server

→ server sends back data (often html, but does not have to be!)

→ browser displays returned data

## Sending Data


→ can also send data to a web server  → e.g. user registration data 

→ different methods/verbs  → e.g. `POST` 

→ specific "path" on web server we need to send the data to (specific `URL`)

&emsp;→ data is attached when request is sent by browser

→ web server receives this data and does something with it

&emsp;→ and usually returns a response of some kind

## In general…

→ web servers listen for incoming requests

→ request contains

&emsp;&emsp;&emsp;→ method  `GET`, `POST`, … 

&emsp;&emsp;&emsp;→ URL  → specifies exactly what we are trying to "access" 

&emsp;&emsp;&emsp;→ `query arguments` (maybe)

&emsp;&emsp;&emsp;→ "attached" `data` (maybe)

the set of what URLs, query arguments, methods and data a web server understands

&emsp;&emsp;&emsp;→ is essentially an API

→ data is not necessarily HTML – can be `JSON`, `XML`, …

## REST APIs

→ REST APIs are special types of APIs

&emsp;→ REST has to do with how they are implemented and their behavior

&emsp;→ as users of the API we don't actually care if it's REST or something else!

→ one of the fundamental characteristics of a REST API is that calls are independent of each other (stateless)

&emsp;→ call to API does not rely on remembering how you interacted with it in the past

&emsp;→ not quite the same with web sites

&emsp;&emsp;&emsp;→ log in

&emsp;&emsp;&emsp;→ now you can access pages on the site

&emsp;&emsp;&emsp;&emsp;&emsp;→ web server remembers who you are

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;→ stateful

## Authentication / Authorization

→ REST APIs are generally secured

&emsp;→ you need to be `authenticated`  → web server needs to know `who you are `

&emsp;&emsp;→ usually a `secret token `you pass in the request

&emsp;&emsp;&emsp;&emsp;→ in something called `headers`

&emsp;&emsp;&emsp;&emsp;&emsp;→ just an extra "bucket" of key-value data that can be

&emsp;&emsp;&emsp;&emsp;&emsp;sent/received along with request

→ you also need to be `authorized `to perform the request

&emsp;→ you may be authorized to read some data

&emsp;→ but you may not be authorized to create/delete that data

&emsp;`Authentication` → establishes who you are to the system you are interacting with

&emsp;`Authorization` → governs what you can/cannot do in system

## API Data Formats

## Resources

→ REST APIs allow us to interact with entities, called resources

```text
→ bank account                          → customer
    → create new account                    → create new customer
    → list accounts for specific customer   → get customer info
    → get balance                           → update customer info
    → deposit, withdraw                     → delete customer
    → delete the account
```


```python
GET https://.../customer/12345/account/5523?query=balance

→ {"balance": 2123.45, "asOf": "2020-04-05T15:35:45+00:00"}
```


```python
POST https://.../customer/12345/account/5523
     + {"action": "withdraw", "amount": 100.0}

→ {"balance": 2023.45, "asOf": "2020-04-05T16:00:00+00:00"}
```

## API Methods

→ since humans design/write these APIs, things are not always consistent!

| Method | Description |
|---------|-------------|
| GET | → retrieves resource(s) |
| GET | → often used with query args |
| POST | → used to create a resource<br>→ issuing the same POST request twice can end up creating two resources |
| PUT, PATCH | → usually used for updating an existing resource |
| DELETE | → delete a resource |

## Status Codes

→ making an HTTP request (GET, POST, etc) always returns a status code

&emsp;→ plus whatever else the API specifies

`2xx` → success
|  | ||
|-------------|---------|-------------|
| 200 | `OK` | request was successful |
| 201 | `Created `| resource created successfully |
| 202 | `Accepted` | request accepted, but not finished processing (async) |

`4XX` → you did something wrong 

|  | ||
|-------------|---------|-------------|
| 400 | `Bad Request` | server did not understand the request |
| 401 | `Unauthorized `| technically this means "not authenticated" |
| 403 | `Forbidden` | this means not authorized |
| 404 | `Not Found `| server cannot find specified resource |

`5XX` → Server had an issue → usually not your fault! 

→ many more…  https://en.wikipedia.org/wiki/List_of_HTTP_status_codes

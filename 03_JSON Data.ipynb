# JSON Data

→ `J`ava`S`cript `O`bject `N`otation

→ it is a `simple` way of representing objects using just strings

&emsp;→ very easy to transmit strings

&emsp;→ over a network, as a text file, etc

→ JSON is a lightweight `standard` that we can use to

| | |
|---|---|
| → `encode` an object into a string | → `serialization` |
| → `decode` a JSON string into an object | → `deserialization` |

→ most often used when transmitting data over the web (e.g. REST APIs)

→ JSON is very simple

&emsp;→ easy for humans to read and write JSON

&emsp;→ easy for computers to parse and generate

&emsp;→ it is a pure text format

&emsp;→ language independent  (Python, C++, C#, JavaScript, Java, etc) 

consists of:

| | |
|---|---|
| object | → `unordered` `key:value` pairs delimited by `{ }` &emsp; (`dictionary`) |
| array  |  → `ordered` list elements separated by `,` and delimited by `[ ]`  &emsp;  (`list`) |
| values | → numbers (integer or with decimal point)<br>→ strings, delimited by double quotes `"…"`<br>→ boolean `true` or `false` (note the `lowercase`!)<br>→ `null`  (`None`)<br>→ object  → so objects can contain other objects, arrays<br>→ array → arrays can contain other arrays, objects |

→ basically JSON looks like a Python dictionary!

→ a JSON object has a `single root object` – everything else is `nested` within it

## Example

```text
''' ←-------------------------------------- it's a string
{ ←---------------------------------------- root is an object
    "firstName": "Eric",←------------------ key: value pairs (key must be a string), (strings must be double-quote delimited)
    "lastName": "Smith",                  
    "address": { ←------------------------- value is another object
        "country": "USA",
        "state": "New York"
    },
    "age": 28,
    "favoriteNumbers": [42, 3.14],←--------- value is a list
    "likesSushi": false,
    "driversLicense": null                                          
}
'''
```

**Important:** order of key:value pairs is irrelevant in JSON - don't count on it!

→ white spaces (spaces, tabs, newlines) do not matter

```python
'''
{
  "firstName": "Eric",
  "lastName": "Smith"
}
'''
```

```python
'''{"firstName":"Eric","lastName":"Smith"}'''
```

→ but which is more human readable?

→ note the stylistic difference: `camelCase` vs `snake_case`

&emsp;→ of course, they are just strings, so you can use whatever you want

## Deserializing JSON (decoding)

→ Python standard library `json` module

&emsp;→ `json.loads(json_string)`

&emsp;&emsp;&emsp;→ `parses` a json string and returns a `dict` object

Since JSON is a standard, Python's `loads` can handle any standard JSON object

## Serializing JSON (encoding)

→ `json.dumps(dict)`  → returns a JSON string 

→ have to be more careful here

→ basic JSON data types are very simple: `int`, `float`, `str`, `bool`, `None`

&emsp;→ Python has a far richer set of data types

&emsp;&emsp;→ `datetime`, `Decimal`, custom classes, etc

&emsp;&emsp;→ those are not serialized by default, and if we try, we'll get an exception

&emsp;&emsp;→ there is a way to specify custom encoders

&emsp;&emsp;&emsp;&emsp;&emsp;→ beyond the scope of this course

### JSON Data

Let's start with a simple JSON object - note that it is just a string!

````python
json_str = '''
{
    "name": "Eric Smith",
    "age": 32,
    "phoneNumbers": [
        {
            "type": "home",
            "number": "(212) 555-3276"
        },
        {
            "type": "work",
            "number": "(332) 555-1234"
        }
    ],
    "spouse": null,
    "children": [],
    "employed": true
}
'''
````

Apart from a few details (such as `true`, `null`), this really just looks like a Python dictionary.

We can in fact **decode** (**deserialize**) this string (this JSON object) into a Python dictionary using the `loads` function in the `json` module:

````python
import json
````

````python
eric = json.loads(json_str)
````

````python
eric
````

This operation is in fact reversable, and we can also **serialize** (**encode**) this Python dictionary into a JSON object using the `dumps` function:

````python
json_str_2 = json.dumps(eric)
````

````python
json_str_2
````

You'll notice that the white space here is very different from what we started with - to JSON whitespace is irrelevant, and in fact, the less whitespace you have in the JSON object the less characters it needs to transmit - but for human reading it can be tough.

We can tell the `dumps` function to use some whitespace to make our (human) life easier:

````python
print(json.dumps(eric, indent=2))
````

The JSON basic value types are very limited: numbers, strings, booleans and null (as well as objects and lists), so many data types in Python cannot just be serialized:

````python
from datetime import datetime

d = {
    "name": "Isaac Newton",
    "dob": datetime(1643, 1, 4)
}
````

````python
d
````

````python
try:
    json.dumps(d)
except TypeError as ex:
    print('TypeError:', ex)
````

As you can see, Python was unable to serialize that datetime object.

There are different ways we can specify custom encoders, but these are mostly beyond the scope of this course (they rely on something called inheritance).

However, there is a way to specify a simple custom encoder, using the named argument `default` in the `dumps` function.

This argument can be used to specify a function that will get called when the default encoder cannot serialize the object. That function should either return the encoded value, or raise a TypeError...

````python
def my_encoder(obj):
    print(f'my_encoder({obj}) called...')
    if isinstance(obj, datetime):
        return obj.isoformat()
    raise TypeError  # only handles datetimes
````

````python
json.dumps(d, default=my_encoder)
````

So `default` is a simple way for you to specify a custom encoder.

It could be used to encode more than just one data type:

````python
from decimal import Decimal
from datetime import date
````

````python
d = {
    "symbol": "IBM",
    "date": date(2020, 9, 21),
    "day": {
        "open": Decimal('120.48'),
        "high": Decimal('120.70'),
        "low": Decimal('118.58'),
        "close": Decimal('120.25'),
        "volume": 5_205_413
    }
}
````

Here we'll have to handle `date` and `Decimal` types ourselves:

````python
def stock_encoder(obj):
    if isinstance(obj, date):
        return obj.isoformat()
    if isinstance(obj, Decimal):
        return str(obj)
    raise TypeError
````

````python
print(json.dumps(d, default=stock_encoder, indent=2))
````

I chose to encode `Decimal` objects as strings, but we could also use floats, rounded to 2 digits after the decimal point:

````python
def stock_encoder(obj):
    if isinstance(obj, date):
        return obj.isoformat()
    if isinstance(obj, Decimal):
        return round(float(obj), 2)
    raise TypeError
````

````python
print(json.dumps(d, default=stock_encoder, indent=2))
````
